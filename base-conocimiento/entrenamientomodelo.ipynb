### 3.1 Selección de Características (Features)

**Objetivo:** Seleccionar las 4 variables más relevantes para predecir SalePrice basándonos en el análisis de correlación previo.

```python
# Importar librerías adicionales necesarias
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor

# Revisar nuevamente las correlaciones con SalePrice
print("="*80)
print("CORRELACIONES CON SALEPRICE")
print("="*80)
correlaciones = df.corr()['SalePrice'].sort_values(ascending=False)
print(correlaciones)
print()

# Seleccionar las 4 variables con mayor correlación (excluyendo SalePrice)
# Basándonos en el análisis del heatmap de la Fase 2
features_seleccionadas = ['OverallQual', 'MasVnrArea', 'LotFrontage']

# Agregar una cuarta variable numérica disponible
# Si el dataset tiene más variables numéricas, seleccionamos la siguiente mejor correlacionada
variables_numericas = df.select_dtypes(include=['float64', 'int64']).columns.tolist()
variables_numericas.remove('SalePrice')  # Excluir la variable objetivo

# Encontrar la cuarta variable con mejor correlación que no esté ya seleccionada
for var in correlaciones.index:
    if var != 'SalePrice' and var not in features_seleccionadas and var in variables_numericas:
        features_seleccionadas.append(var)
        break

# Si no hay suficientes variables, usar las disponibles
if len(features_seleccionadas) < 4:
    for var in variables_numericas:
        if var not in features_seleccionadas:
            features_seleccionadas.append(var)
            if len(features_seleccionadas) == 4:
                break

print("="*80)
print("VARIABLES SELECCIONADAS PARA EL MODELO")
print("="*80)
for i, feature in enumerate(features_seleccionadas, 1):
    corr_value = correlaciones[feature]
    print(f"{i}. {feature:20s} - Correlación: {corr_value:.4f}")
print()

# Definir X (variables independientes) y y (variable dependiente)
X = df[features_seleccionadas]
y = df['SalePrice']

print(f"Forma de X (features): {X.shape}")
print(f"Forma de y (target): {y.shape}")
```

   - **Lógica de negocio:** La calidad general de construcción y acabados es uno de los factores más determinantes en el precio de una propiedad. Casas con mejor calidad tienen materiales superiores, mejor diseño y mayor durabilidad.
   - **Justificación:** Variable categórica ordinal que captura la percepción general de calidad.

2. **MasVnrArea (Área de Revestimiento de Mampostería)**
   - **Lógica estadística:** Correlación moderada-fuerte con SalePrice
   - **Lógica de negocio:** El revestimiento de mampostería (ladrillo, piedra) es un indicador de calidad y estética. Propiedades con mayor área de revestimiento suelen ser más atractivas y valiosas.
   - **Justificación:** Variable continua que refleja acabados premium.

3. **LotFrontage (Frente del Lote)**
   - **Lógica estadística:** Correlación positiva moderada con SalePrice
   - **Lógica de negocio:** El tamaño del frente del lote afecta la privacidad, accesibilidad y percepción de espacio. Lotes más grandes con mayor frente son más deseables.

### 4.1 Realizar Predicciones

**Objetivo:** Usar los modelos entrenados para predecir SalePrice en el conjunto de prueba.

```python
# Realizar predicciones con ambos modelos
print("="*80)
print("REALIZANDO PREDICCIONES")
print("="*80)
print()

# Predicciones del Modelo A (Regresión Lineal)
y_pred_lineal = modelo_lineal.predict(X_test)
print("✅ Predicciones del Modelo A (Regresión Lineal) completadas")

# Predicciones del Modelo B (Random Forest)
y_pred_rf = modelo_random_forest.predict(X_test)
print("✅ Predicciones del Modelo B (Random Forest) completadas")
print()

# Mostrar algunas predicciones vs valores reales
print("Comparación de predicciones (primeras 10 muestras):")
print("="*80)
print(f"{'Real':>15s} {'Pred. Lineal':>15s} {'Pred. RF':>15s} {'Error Lineal':>15s} {'Error RF':>15s}")
print("="*80)

for i in range(min(10, len(y_test))):
    real = y_test.iloc[i]
    pred_lin = y_pred_lineal[i]
    pred_rf = y_pred_rf[i]
    error_lin = abs(real - pred_lin)
    error_rf = abs(real - pred_rf)
    
    print(f"${real:>14,.0f} ${pred_lin:>14,.0f} ${pred_rf:>14,.0f} ${error_lin:>14,.0f} ${error_rf:>14,.0f}")
```

### 4.2 Cálculo de Métricas de Evaluación

**Objetivo:** Calcular RMSE y R² para ambos modelos y compararlos.

```python
# Calcular métricas para Modelo A (Regresión Lineal)
rmse_lineal = np.sqrt(mean_squared_error(y_test, y_pred_lineal))
r2_lineal = r2_score(y_test, y_pred_lineal)

# Calcular métricas para Modelo B (Random Forest)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf = r2_score(y_test, y_pred_rf)

# Mostrar resultados
print("="*80)
print("MÉTRICAS DE EVALUACIÓN")
print("="*80)
print()

print("MODELO A: REGRESIÓN LINEAL")
print("-" * 40)
print(f"  RMSE (Error Cuadrático Medio): ${rmse_lineal:,.2f}")
print(f"  R² (Coeficiente de Determinación): {r2_lineal:.4f} ({r2_lineal*100:.2f}%)")
print()

print("MODELO B: RANDOM FOREST REGRESSOR")
print("-" * 40)
print(f"  RMSE (Error Cuadrático Medio): ${rmse_rf:,.2f}")
print(f"  R² (Coeficiente de Determinación): {r2_rf:.4f} ({r2_rf*100:.2f}%)")
print()

# Comparación
print("="*80)
print("COMPARACIÓN DE MODELOS")
print("="*80)
print()

mejor_rmse = "Regresión Lineal" if rmse_lineal < rmse_rf else "Random Forest"
mejor_r2 = "Regresión Lineal" if r2_lineal > r2_rf else "Random Forest"

print(f"Diferencia en RMSE: ${abs(rmse_lineal - rmse_rf):,.2f}")
print(f"  ✅ Mejor RMSE (menor error): {mejor_rmse}")
print()

print(f"Diferencia en R²: {abs(r2_lineal - r2_rf):.4f}")
print(f"  ✅ Mejor R² (mayor explicación): {mejor_r2}")
print()

# Determinar el mejor modelo general
if mejor_rmse == mejor_r2:
    print(f"🏆 MODELO GANADOR: {mejor_rmse}")
    print(f"   Este modelo tiene mejor rendimiento en ambas métricas.")
else:
    print("⚖️  RESULTADO MIXTO:")
    print(f"   - Mejor RMSE: {mejor_rmse}")
    print(f"   - Mejor R²: {mejor_r2}")
    print(f"   La elección depende de la prioridad (minimizar error vs maximizar explicación)")
```


#### ¿Qué significa RMSE (Root Mean Squared Error)?

- **Definición:** Es la raíz cuadrada del promedio de los errores al cuadrado.
- **Unidades:** Se expresa en las mismas unidades que la variable objetivo (dólares en este caso).
- **Interpretación:** Nos dice cuánto nos equivocamos en promedio al predecir el precio.
  - Ejemplo: Si RMSE = $50,000, significa que en promedio nuestras predicciones se desvían $50,000 del precio real.
- **Objetivo:** **Minimizar** (valores más bajos son mejores).
- **Ventaja:** Penaliza más los errores grandes, lo cual es útil cuando queremos evitar predicciones muy alejadas.

#### ¿Qué significa R² (Coeficiente de Determinación)?

- **Definición:** Proporción de la varianza en la variable dependiente que es predecible a partir de las variables independientes.
- **Rango:** De 0 a 1 (o 0% a 100%).
- **Interpretación:** Nos dice qué porcentaje de la variabilidad del precio es explicado por nuestro modelo.
  - R² = 0.70 (70%) → El modelo explica el 70% de la variación en los precios.
  - R² = 0.90 (90%) → El modelo explica el 90% de la variación en los precios.
- **Objetivo:** **Maximizar** (valores más altos son mejores).
- **Interpretación práctica:**
  - R² < 0.3 (30%): Modelo pobre
  - R² 0.3-0.5 (30-50%): Modelo aceptable
  - R² 0.5-0.7 (50-70%): Modelo bueno
  - R² > 0.7 (>70%): Modelo muy bueno
  - R² > 0.9 (>90%): Modelo excelente
